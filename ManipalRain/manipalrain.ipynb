```python
import pandas as pd
import numpy as np
import datetime

df = pd.read_csv('ManipalRains.csv')
np.random.seed(0) 
# random.seed() is used to set the seed for the random number generator in NumPy. By setting a seed value, you can ensure that the random numbers generated are reproducible, meaning that you will get the same sequence of random numbers each time you run the code. This is useful for debugging and testing purposes, as it allows you to compare results across different runs of the code.
df.head()
df.tail()
#tail() is used to display the last few rows of the DataFrame. By default, it displays the last 5 rows, but you can specify a different number of rows to display by passing an integer as an argument.

#head() is used to display the first few rows of the DataFrame. By default, it displays the first 5 rows, but you can specify a different number of rows to display by passing an integer as an argument.
#isnull() is used to detect missing values in the DataFrame. It returns a DataFrame of the same shape as the original, with True for missing values and False for non-missing values. The sum() function is then used to count the number of missing values in each column, and the second sum() counts the total number of missing values in the entire DataFrame.
#loc[] is used to access a group of rows and columns by labels or a boolean array. In this case, it is used to select all rows and the columns from 'MinTemp' to 'Evaporation', and then the head() function is called to display the first few rows of this subset of the DataFrame.

print("Original:", df.isnull().sum().sum(), "missing values")
print("small subset of the data with missing values:", df.loc[:, 'MinTemp':'Evaporation'].head())

ds=df.dropna()
fs=df.dropna(axis=1)

#dropna is used to remove missing values from the DataFrame. By default, it removes any row that contains at least one missing value. The parameter axis=1 can be used to remove columns instead of rows.
#shape is an attribute of the DataFrame that returns a tuple representing the dimensions of the DataFrame. The first element of the tuple is the number of rows, and the second element is the number of columns.

print("Original DataFrame:", df.shape[0], "rows,", df.shape[1], "columns")
print("DataFrame after dropping rows with missing values:", ds.shape[0], "rows,", ds.shape[1], "columns")
print("DataFrame after dropping columns with missing values:", fs.shape[0], "rows,", fs.shape[1], "columns")

filled=df.fillna(df.mean())
#fillna is used to fill missing values in the DataFrame. In this case, it is used to fill missing values with the mean of each column. The mean() function calculates the mean of each column, and the fillna() function replaces the missing values with the corresponding mean value.       
#mean(0) calculates the mean along the columns (axis=0), which means it computes the mean for each column separately. it is the dwfault behavior of the mean() function, so you can also use mean() without any arguments to achieve the same result.
#mean(1),etc. would calculate the mean along the rows (axis=1), which means it computes the mean for each row separately.
#axis=0 is the default behavior for many pandas functions, including mean(), which means that the operation is performed column-wise. When you specify axis=1, the operation is performed row-wise.

print("DataFrame after filling missing values with mean:", filled.shape[0], "rows,", filled.shape[1], "columns")

# only for shape 0 is rows and 1 is columns.

print("The new DataFrame after filling missing values with mean:", filled.head())
# this is an example of mean imputation, which is a common technique for handling missing data. By replacing missing values with the mean of the respective column, we can maintain the overall distribution of the data while avoiding the loss of information that occurs when dropping rows or columns with missing values.

print("The new dataframe method:\n", df.fillna(method='bfill').head())
# this is an example of backfill imputation, which is another technique for handling missing data. By replacing missing values with the next valid observation in the column, we can maintain the overall structure of the data while avoiding the loss of information that occurs when dropping rows or columns with missing values.

print(ds.dtypes())
```
