*Cleaned and Preprocessed Data*

```python
import seaborn as sns
import pandas as pd
import numpy as np

data=pd.read_csv("ManipalRains.csv")

print("The number of rows:", data.shape[0], "The number of columns:", data.shape[1])

data["Date"] = pd.to_datetime(data["Date"],format="%d-%m-%Y")
data = data.sort_values("Date")

data['Rainfall'].fillna(0, inplace=True)
wd_columns = ["WindGustDir", "WindDir9am", "WindDir3pm"]
for i in wd_columns:
    data[i] = data[i].fillna(data[i].mode()[0])
data.drop(columns=["Evaporation", "Sunshine"], inplace=True)
data.dropna(subset=["RainToday", "RainTomorrow"], inplace=True)

n_columns=["MinTemp","MaxTemp","WindGustSpeed","Cloud9am", "Cloud3pm","Temp3pm","Temp9am", "Humidity9am","Humidity3pm","Pressure9am","Pressure3pm","WindSpeed9am","WindSpeed3pm"]
data[n_columns]=data[n_columns].interpolate(method="linear", limit_direction="both")

print(data.isnull().sum())











#inplace=True is used to modify the original DataFrame directly, rather than creating a new DataFrame with the missing values filled. This means that the changes made by fillna() will be applied to the original DataFrame, and the original DataFrame will be updated with the filled values.
# if inplace=False (default), a new DataFrame with the filled values will be returned, and the original DataFrame will remain unchanged.
# by default, fillna() fills missing values with NaN (Not a Number) in the DataFrame. However, you can specify a different value to fill the missing values by passing it as an argument to the fillna() function. In this case, we 
# by dault inplace=False, fillna() returns a new DataFrame with the missing values filled, and the original DataFrame remains unchanged. If inplace=True is specified, the original DataFrame is modified directly, and no new DataFrame is returned.
#the dataframe here is called data, and the missing values in the DataFrame are filled with 0 using the fillna() function. The inplace=True argument is used to modify the original DataFrame directly, rather than creating a new DataFrame with the filled values. This means that the changes made by fillna() will be applied to the original DataFrame, and the original DataFrame will be updated with the filled values.
# data['Rainfall'] = data['Rainfall'].fillna(0) is used to fill the missing values in the 'Rainfall' column of the DataFrame with 0. The fillna() function is called on the 'Rainfall' column, and the inplace=True argument is not specified, so a new Series with the filled values is returned. This new Series is then assigned back to the 'Rainfall' column of the DataFrame, effectively updating the original DataFrame with the filled values.


```
